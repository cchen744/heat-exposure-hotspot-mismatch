```python
import importlib

packages_to_check = [
    "pandas",
    "geopandas",
    "matplotlib",
    "requests"
]

check_results = []

for pkg in packages_to_check:
    try:
        mod = importlib.import_module(pkg)
        version = getattr(mod, "__version__", "unknown")
        check_results.append({"package": pkg, "installed": True, "version": version})
    except Exception as e:
        check_results.append({"package": pkg, "installed": False, "version": str(e)})

pd.DataFrame(check_results)
```
