```python
# =========================================================
# 10_alphaearth_rawdim_model.ipynb
# Compare:
# 1) baseline factors only
# 2) baseline + PC1-PC7
# 3) baseline + stable raw AlphaEarth dims
# 4) baseline + PC1-PC7 + stable raw dims
# =========================================================

from pathlib import Path
import re
import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

from sklearn.model_selection import KFold, cross_validate
from sklearn.ensemble import RandomForestRegressor

# -----------------------------
# paths
# -----------------------------
ROOT = Path("/Users/yufeizhou/Desktop/heat-exposure-compare")
OUT = ROOT / "outputs/pilot/alphaearth"
OUT.mkdir(parents=True, exist_ok=True)

CITY_CFG = {
    "Houston": {
        "embed_clean": OUT / "houston_alphaearth_2024_tract_mean_mosaic_clean.csv",
        "heat_file": ROOT / "data_processed/houston/houston_master_with_lst_hi_fixed.gpkg",
        "pc_file": OUT / "houston_alphaearth_pc7_scores_corrected.csv",
    },
    "Phoenix": {
        "embed_clean": OUT / "phoenix_alphaearth_2024_tract_mean_mosaic_clean.csv",
        "heat_file": ROOT / "data_processed/phoenix/phoenix_master_with_lst_hi_fixed.gpkg",
        "pc_file": OUT / "phoenix_alphaearth_pc7_scores_corrected.csv",
    },
}

WINNER_FILE = OUT / "city_top4_pc_factor_stability_winners.csv"

# -----------------------------
# helpers
# -----------------------------
def normalize_geoid(series: pd.Series) -> pd.Series:
    s = series.astype(str).str.strip()
    s = s.str.replace(".0", "", regex=False)
    s = s.str.extract(r"(\d+)")[0].fillna(series.astype(str))
    return s.astype(str).str.zfill(11)

def find_col(cols, candidates, required=True):
    lower_map = {str(c).lower(): c for c in cols}
    for cand in candidates:
        if cand.lower() in lower_map:
            return lower_map[cand.lower()]
    if required:
        raise KeyError(f"Cannot find any of {candidates} in columns: {list(cols)}")
    return None

def parse_dims_from_text(x):
    if pd.isna(x):
        return []
    return re.findall(r"A\d{2}", str(x))

def prepare_heat_df(city, heat_file):
    gdf = gpd.read_file(heat_file)
    heat_df = gdf.drop(columns="geometry", errors="ignore").copy()

    geoid_col = find_col(heat_df.columns, ["GEOID", "geoid"])
    heat_df["GEOID"] = normalize_geoid(heat_df[geoid_col])

    lst_col = find_col(heat_df.columns, ["lst_c", "lst"])
    hi_col  = find_col(heat_df.columns, ["hi_c", "hi"])
    income_col = find_col(heat_df.columns, ["median_household_income_clean", "median_household_income"])
    poverty_col = find_col(heat_df.columns, ["poverty_rate"])
    pop_col = find_col(heat_df.columns, ["total_population_clean", "total_population"])
    aland_col = find_col(heat_df.columns, ["ALAND"])
    awater_col = find_col(heat_df.columns, ["AWATER"])

    # numeric
    for c in [lst_col, hi_col, income_col, poverty_col, pop_col, aland_col, awater_col]:
        heat_df[c] = pd.to_numeric(heat_df[c], errors="coerce")

    # target: hi_minus_lst
    if "hi_minus_lst" not in heat_df.columns:
        heat_df["hi_minus_lst"] = heat_df[hi_col] - heat_df[lst_col]
    else:
        heat_df["hi_minus_lst"] = pd.to_numeric(heat_df["hi_minus_lst"], errors="coerce")

    # z-scored target
    if "gap_z_hi_minus_lst" not in heat_df.columns:
        mu = heat_df["hi_minus_lst"].mean()
        sd = heat_df["hi_minus_lst"].std()
        heat_df["gap_z_hi_minus_lst"] = (heat_df["hi_minus_lst"] - mu) / sd
    else:
        heat_df["gap_z_hi_minus_lst"] = pd.to_numeric(heat_df["gap_z_hi_minus_lst"], errors="coerce")

    info = {
        "lst_col": lst_col,
        "hi_col": hi_col,
        "income_col": income_col,
        "poverty_col": poverty_col,
        "pop_col": pop_col,
        "aland_col": aland_col,
        "awater_col": awater_col,
        "target_col": "gap_z_hi_minus_lst",
    }

    return heat_df, info

def prepare_embed_df(embed_file):
    df = pd.read_csv(embed_file, dtype={"GEOID": str}).copy()
    geoid_col = find_col(df.columns, ["GEOID", "geoid"])
    df["GEOID"] = normalize_geoid(df[geoid_col])

    dim_cols = sorted([c for c in df.columns if re.fullmatch(r"A\d{2}", str(c))])
    for c in dim_cols:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    return df, dim_cols

def prepare_pc_df(pc_file):
    df = pd.read_csv(pc_file, dtype={"GEOID": str}).copy()
    geoid_col = find_col(df.columns, ["GEOID", "geoid"])
    df["GEOID"] = normalize_geoid(df[geoid_col])

    pc_cols = [c for c in ["PC1", "PC2", "PC3", "PC4", "PC5", "PC6", "PC7"] if c in df.columns]
    for c in pc_cols:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    return df, pc_cols

def get_city_stable_dims(winner_df, city, available_dim_cols):
    city_col = find_col(winner_df.columns, ["city"])
    dims_col = find_col(winner_df.columns, ["example_top_dims"])

    sub = winner_df[winner_df[city_col].astype(str).str.lower() == city.lower()].copy()

    dims = []
    for x in sub[dims_col]:
        dims.extend(parse_dims_from_text(x))

    dims = sorted(set(dims))
    dims = [d for d in dims if d in available_dim_cols]
    return dims

def fit_cv_models(df_model, feature_sets, target_col, city):
    rows = []
    fold_rows = []

    cv = KFold(n_splits=5, shuffle=True, random_state=42)
    scoring = {
        "r2": "r2",
        "rmse": "neg_root_mean_squared_error",
        "mae": "neg_mean_absolute_error",
    }

    for model_name, feat_cols in feature_sets.items():
        X = df_model[feat_cols].copy()
        y = df_model[target_col].copy()

        rf = RandomForestRegressor(
            n_estimators=500,
            random_state=42,
            min_samples_leaf=2,
            n_jobs=-1,
        )

        cv_out = cross_validate(
            rf,
            X,
            y,
            cv=cv,
            scoring=scoring,
            return_train_score=False,
            n_jobs=1,
        )

        row = {
            "city": city,
            "model": model_name,
            "n_features": len(feat_cols),
            "features": ", ".join(feat_cols),
            "cv_r2_mean": np.mean(cv_out["test_r2"]),
            "cv_r2_std": np.std(cv_out["test_r2"]),
            "cv_rmse_mean": -np.mean(cv_out["test_rmse"]),
            "cv_rmse_std": np.std(-cv_out["test_rmse"]),
            "cv_mae_mean": -np.mean(cv_out["test_mae"]),
            "cv_mae_std": np.std(-cv_out["test_mae"]),
        }
        rows.append(row)

        for i in range(len(cv_out["test_r2"])):
            fold_rows.append({
                "city": city,
                "model": model_name,
                "fold": i + 1,
                "r2": cv_out["test_r2"][i],
                "rmse": -cv_out["test_rmse"][i],
                "mae": -cv_out["test_mae"][i],
            })

    return pd.DataFrame(rows), pd.DataFrame(fold_rows)

# -----------------------------
# read winners
# -----------------------------
winner_df = pd.read_csv(WINNER_FILE).copy()

all_summary = []
all_folds = []
all_dim_tables = []

for city, cfg in CITY_CFG.items():
    print(f"\n===== {city} =====")
    print("embed file:", cfg["embed_clean"])
    print("heat file :", cfg["heat_file"])
    print("pc file   :", cfg["pc_file"])

    embed_df, all_dim_cols = prepare_embed_df(cfg["embed_clean"])
    heat_df, info = prepare_heat_df(city, cfg["heat_file"])
    pc_df, pc_cols = prepare_pc_df(cfg["pc_file"])

    stable_dims = get_city_stable_dims(winner_df, city, all_dim_cols)

    print("n embed dims:", len(all_dim_cols))
    print("stable raw dims selected:", stable_dims)

    dim_table = pd.DataFrame({
        "city": city,
        "selected_raw_dim": stable_dims
    })
    all_dim_tables.append(dim_table)

    use_heat_cols = [
        "GEOID",
        info["target_col"],
        info["income_col"],
        info["poverty_col"],
        info["pop_col"],
        info["aland_col"],
        info["awater_col"],
    ]

    merged = (
        heat_df[use_heat_cols].copy()
        .merge(embed_df[["GEOID"] + stable_dims], on="GEOID", how="inner")
        .merge(pc_df[["GEOID"] + pc_cols], on="GEOID", how="left")
    )

    # one common modeling sample for fair comparison
    baseline_cols = [
        info["income_col"],
        info["poverty_col"],
        info["pop_col"],
        info["aland_col"],
        info["awater_col"],
    ]

    all_needed = [info["target_col"]] + baseline_cols + stable_dims + pc_cols
    df_model = merged[["GEOID"] + all_needed].copy()

    for c in all_needed:
        df_model[c] = pd.to_numeric(df_model[c], errors="coerce")

    df_model = df_model.dropna(subset=all_needed).copy()

    print("merged rows:", len(merged))
    print("model rows after common dropna:", len(df_model))

    feature_sets = {
        "baseline_factors_only": baseline_cols,
        "baseline_plus_pc1_to_pc7": baseline_cols + pc_cols,
        "baseline_plus_stable_rawdims": baseline_cols + stable_dims,
        "baseline_plus_pc1_to_pc7_plus_stable_rawdims": baseline_cols + pc_cols + stable_dims,
    }

    summary_df, fold_df = fit_cv_models(
        df_model=df_model,
        feature_sets=feature_sets,
        target_col=info["target_col"],
        city=city,
    )

    # add delta vs baseline
    base_r2 = summary_df.loc[summary_df["model"] == "baseline_factors_only", "cv_r2_mean"].iloc[0]
    base_rmse = summary_df.loc[summary_df["model"] == "baseline_factors_only", "cv_rmse_mean"].iloc[0]
    base_mae = summary_df.loc[summary_df["model"] == "baseline_factors_only", "cv_mae_mean"].iloc[0]

    summary_df["delta_r2_vs_baseline"] = summary_df["cv_r2_mean"] - base_r2
    summary_df["delta_rmse_vs_baseline"] = summary_df["cv_rmse_mean"] - base_rmse
    summary_df["delta_mae_vs_baseline"] = summary_df["cv_mae_mean"] - base_mae
    summary_df["n_model_rows"] = len(df_model)

    all_summary.append(summary_df)
    all_folds.append(fold_df)

# -----------------------------
# combine + save
# -----------------------------
summary_all = pd.concat(all_summary, ignore_index=True)
folds_all = pd.concat(all_folds, ignore_index=True)
dims_all = pd.concat(all_dim_tables, ignore_index=True)

summary_csv = OUT / "city_rawdim_vs_pc_model_summary.csv"
folds_csv = OUT / "city_rawdim_vs_pc_model_folds.csv"
dims_csv = OUT / "city_rawdim_selected_dims.csv"

summary_all.to_csv(summary_csv, index=False)
folds_all.to_csv(folds_csv, index=False)
dims_all.to_csv(dims_csv, index=False)

print("\nSaved:")
print(summary_csv)
print(folds_csv)
print(dims_csv)

print("\n===== Summary table =====")
display(summary_all)

print("\n===== Selected raw dims =====")
display(dims_all)

# -----------------------------
# compact meeting table
# -----------------------------
meeting_df = summary_all[[
    "city", "model", "n_model_rows", "n_features",
    "cv_r2_mean", "cv_rmse_mean", "cv_mae_mean",
    "delta_r2_vs_baseline", "delta_rmse_vs_baseline", "delta_mae_vs_baseline"
]].copy()

meeting_csv = OUT / "city_rawdim_vs_pc_meeting_table.csv"
meeting_df.to_csv(meeting_csv, index=False)

print("\nSaved meeting table:", meeting_csv)
display(meeting_df)

# -----------------------------
# plot delta R2
# -----------------------------
plot_df = summary_all[summary_all["model"] != "baseline_factors_only"].copy()
plot_df["label"] = plot_df["city"] + "_" + plot_df["model"]

plt.figure(figsize=(10, 5))
plt.bar(plot_df["label"], plot_df["delta_r2_vs_baseline"])
plt.axhline(0, linewidth=1)
plt.ylabel("Delta R² vs baseline")
plt.title("Raw-dimension vs PCA model improvement")
plt.xticks(rotation=20, ha="right")
plt.tight_layout()

plot_png = OUT / "city_rawdim_vs_pc_delta_r2.png"
plt.savefig(plot_png, dpi=300, bbox_inches="tight")
plt.show()

print("Saved plot:", plot_png)

# -----------------------------
# best model per city
# -----------------------------
best_df = summary_all.sort_values(["city", "cv_r2_mean"], ascending=[True, False]).groupby("city").head(1).copy()
best_csv = OUT / "city_rawdim_vs_pc_best_model_by_city.csv"
best_df.to_csv(best_csv, index=False)

print("\nSaved best-model table:", best_csv)
display(best_df)
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
from scipy.stats import spearmanr

# =========================
# paths
# =========================
ROOT = Path("/Users/yufeizhou/Desktop/heat-exposure-compare")
OUT = ROOT / "outputs/pilot/alphaearth"
OUT.mkdir(parents=True, exist_ok=True)

city_cfg = {
    "Houston": {
        "embed_file": OUT / "houston_alphaearth_2024_tract_mean_mosaic_clean.csv",
        "heat_file": ROOT / "data_processed/houston/houston_master_with_lst_hi_fixed.gpkg",
    },
    "Phoenix": {
        "embed_file": OUT / "phoenix_alphaearth_2024_tract_mean_mosaic_clean.csv",
        "heat_file": ROOT / "data_processed/phoenix/phoenix_master_with_lst_hi_fixed.gpkg",
    },
}

selected_dims_file = OUT / "city_rawdim_selected_dims.csv"

# =========================
# helpers
# =========================
def pick_col(df, candidates, required=True):
    lower_map = {c.lower(): c for c in df.columns}
    for cand in candidates:
        if cand.lower() in lower_map:
            return lower_map[cand.lower()]
    if required:
        raise KeyError(f"Cannot find any of {candidates} in columns: {list(df.columns)}")
    return None

def normalize_geoid(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip()
    if s.endswith(".0"):
        s = s[:-2]
    s = s.replace(" ", "")
    if s == "" or s.lower() == "nan":
        return np.nan
    # keep digits only
    s = "".join(ch for ch in s if ch.isdigit())
    if s == "":
        return np.nan
    # census tract GEOID should be 11 digits
    return s.zfill(11)

def plot_map(ax, gdf, col, title, cmap="viridis"):
    valid = gdf.dropna(subset=[col]).copy()
    if valid.empty:
        ax.set_title(f"{title}\n(no data)", fontsize=10)
        ax.axis("off")
        return

    vals = pd.to_numeric(valid[col], errors="coerce").replace([np.inf, -np.inf], np.nan).dropna()
    if len(vals) == 0:
        ax.set_title(f"{title}\n(no valid numeric values)", fontsize=10)
        ax.axis("off")
        return

    vmin = np.nanquantile(vals, 0.02)
    vmax = np.nanquantile(vals, 0.98)
    if (not np.isfinite(vmin)) or (not np.isfinite(vmax)) or np.isclose(vmin, vmax):
        vmin = float(vals.min())
        vmax = float(vals.max())

    gdf.plot(ax=ax, color="lightgrey", edgecolor="none")
    valid.plot(
        ax=ax,
        column=col,
        cmap=cmap,
        linewidth=0,
        legend=True,
        vmin=vmin,
        vmax=vmax,
        missing_kwds={"color": "lightgrey"},
    )
    ax.set_title(title, fontsize=10)
    ax.set_axis_off()

# =========================
# load selected dims
# =========================
if not selected_dims_file.exists():
    raise FileNotFoundError(f"selected dims file not found: {selected_dims_file}")

sel = pd.read_csv(selected_dims_file).copy()
city_col = pick_col(sel, ["city"])
dim_col = pick_col(sel, ["selected_raw_dim", "raw_dim", "selected_dim", "dim", "embedding_dim"])

all_corr_rows = []
all_pair_rows = []

priority_vars = [
    "lst_c",
    "hi_c",
    "hi_minus_lst",
    "median_household_income_clean",
    "poverty_rate",
    "ALAND",
    "AWATER",
]

top_n_map = 4

for city, cfg in city_cfg.items():
    print(f"\n===== {city} =====")

    embed_file = Path(cfg["embed_file"])
    heat_file = Path(cfg["heat_file"])

    if not embed_file.exists():
        raise FileNotFoundError(f"{city} embed file not found: {embed_file}")
    if not heat_file.exists():
        raise FileNotFoundError(f"{city} heat file not found: {heat_file}")

    embed_df = pd.read_csv(embed_file).copy()
    gdf = gpd.read_file(heat_file).copy()

    embed_geoid = pick_col(embed_df, ["GEOID", "geoid"])
    heat_geoid = pick_col(gdf, ["GEOID", "geoid"])

    embed_df["GEOID"] = embed_df[embed_geoid].apply(normalize_geoid)
    gdf["GEOID"] = gdf[heat_geoid].apply(normalize_geoid)

    embed_df = embed_df.dropna(subset=["GEOID"]).copy()
    gdf = gdf.dropna(subset=["GEOID"]).copy()

    # remove duplicates if any
    embed_df = embed_df.drop_duplicates(subset=["GEOID"]).copy()
    gdf = gdf.drop_duplicates(subset=["GEOID"]).copy()

    # resolve observable columns flexibly
    lst_col = pick_col(gdf, ["lst_c", "lst_summer_mean_c"], required=False)
    hi_col = pick_col(gdf, ["hi_c", "hi_summer_mean_c"], required=False)
    income_col = pick_col(gdf, ["median_household_income_clean", "median_household_income"], required=False)
    poverty_col = pick_col(gdf, ["poverty_rate"], required=False)
    pop_col = pick_col(gdf, ["total_population_clean", "total_population", "population"], required=False)
    aland_col = pick_col(gdf, ["ALAND"], required=False)
    awater_col = pick_col(gdf, ["AWATER"], required=False)

    rename_map = {}
    if lst_col and lst_col != "lst_c":
        rename_map[lst_col] = "lst_c"
    if hi_col and hi_col != "hi_c":
        rename_map[hi_col] = "hi_c"
    if income_col and income_col != "median_household_income_clean":
        rename_map[income_col] = "median_household_income_clean"
    if poverty_col and poverty_col != "poverty_rate":
        rename_map[poverty_col] = "poverty_rate"
    if pop_col and pop_col != "total_population_clean":
        rename_map[pop_col] = "total_population_clean"
    if aland_col and aland_col != "ALAND":
        rename_map[aland_col] = "ALAND"
    if awater_col and awater_col != "AWATER":
        rename_map[awater_col] = "AWATER"

    if rename_map:
        gdf = gdf.rename(columns=rename_map)

    for col in ["lst_c", "hi_c", "median_household_income_clean", "poverty_rate",
                "total_population_clean", "ALAND", "AWATER"]:
        if col in gdf.columns:
            gdf[col] = pd.to_numeric(gdf[col], errors="coerce")

    if "hi_minus_lst" not in gdf.columns:
        if ("hi_c" in gdf.columns) and ("lst_c" in gdf.columns):
            gdf["hi_minus_lst"] = gdf["hi_c"] - gdf["lst_c"]

    city_dims = (
        sel.loc[sel[city_col] == city, dim_col]
        .dropna()
        .astype(str)
        .drop_duplicates()
        .tolist()
    )

    if len(city_dims) == 0:
        print(f"No selected raw dims found for {city}")
        continue

    city_dims = [d for d in city_dims if d in embed_df.columns]
    if len(city_dims) == 0:
        print(f"No selected raw dims present in embed file for {city}")
        continue

    for d in city_dims:
        embed_df[d] = pd.to_numeric(embed_df[d], errors="coerce")

    # debug GEOID overlap
    embed_set = set(embed_df["GEOID"])
    heat_set = set(gdf["GEOID"])
    overlap = embed_set & heat_set

    print("embed rows:", len(embed_df))
    print("heat rows:", len(gdf))
    print("embed unique GEOID:", len(embed_set))
    print("heat unique GEOID:", len(heat_set))
    print("overlap GEOID:", len(overlap))
    print("embed sample:", embed_df["GEOID"].head(5).tolist())
    print("heat sample:", gdf["GEOID"].head(5).tolist())

    merged = gdf.merge(embed_df[["GEOID"] + city_dims], on="GEOID", how="inner").copy()

    print("merged rows:", len(merged))
    print("selected raw dims:", city_dims)

    available_vars = [v for v in priority_vars if v in merged.columns]
    print("available observable vars:", available_vars)

    if len(merged) == 0:
        print(f"WARNING: {city} merged rows = 0, skip plotting for now")
        continue

    city_corr_rows = []
    for raw_dim in city_dims:
        for var in available_vars:
            sub = merged[[raw_dim, var]].dropna().copy()
            n = len(sub)

            if n < 20:
                r = np.nan
            else:
                r = spearmanr(sub[raw_dim], sub[var]).correlation

            city_corr_rows.append({
                "city": city,
                "raw_dim": raw_dim,
                "best_observable_var_candidate": var,
                "spearman_r": r,
                "abs_spearman_r": np.nan if pd.isna(r) else abs(r),
                "n": n,
            })

    city_corr_df = pd.DataFrame(city_corr_rows)
    all_corr_rows.append(city_corr_df)

    dim_rank_df = (
        city_corr_df.groupby("raw_dim", as_index=False)["abs_spearman_r"]
        .max()
        .sort_values("abs_spearman_r", ascending=False)
        .reset_index(drop=True)
    )

    chosen_dims = dim_rank_df["raw_dim"].head(top_n_map).tolist()

    city_pair_rows = []
    for raw_dim in chosen_dims:
        tmp = (
            city_corr_df.loc[city_corr_df["raw_dim"] == raw_dim]
            .sort_values(["abs_spearman_r", "n"], ascending=[False, False])
            .reset_index(drop=True)
        )
        if len(tmp) == 0:
            continue

        best = tmp.iloc[0]
        city_pair_rows.append({
            "city": city,
            "raw_dim": raw_dim,
            "best_observable_var": best["best_observable_var_candidate"],
            "spearman_r": best["spearman_r"],
            "abs_spearman_r": best["abs_spearman_r"],
            "n": best["n"],
        })

    city_pair_df = pd.DataFrame(city_pair_rows)
    all_pair_rows.append(city_pair_df)

    nrows = len(city_pair_df)
    if nrows == 0:
        print(f"No valid pair rows for {city}")
        continue

    fig, axes = plt.subplots(nrows=nrows, ncols=2, figsize=(12, 4 * nrows))
    if nrows == 1:
        axes = np.array([axes])

    for i, row in city_pair_df.reset_index(drop=True).iterrows():
        raw_dim = row["raw_dim"]
        best_var = row["best_observable_var"]
        r = row["spearman_r"]

        plot_map(
            axes[i, 0],
            merged,
            raw_dim,
            f"{city} raw dim: {raw_dim}",
            cmap="viridis"
        )
        plot_map(
            axes[i, 1],
            merged,
            best_var,
            f"{city} linked var: {best_var}\nSpearman r = {r:.3f}",
            cmap="magma"
        )

    plt.tight_layout()
    panel_png = OUT / f"{city.lower()}_rawdim_observable_pairs.png"
    plt.savefig(panel_png, dpi=300, bbox_inches="tight")
    plt.show()

    print("saved panel:", panel_png)

# =========================
# save outputs
# =========================
if len(all_corr_rows) == 0:
    raise ValueError("No correlation rows were produced.")

corr_df = pd.concat(all_corr_rows, ignore_index=True)

if len(all_pair_rows) == 0:
    raise ValueError("No pair rows were produced.")

pair_df = pd.concat(all_pair_rows, ignore_index=True)

corr_csv = OUT / "city_rawdim_observable_spearman.csv"
pair_csv = OUT / "city_rawdim_pair_summary.csv"

corr_df.to_csv(corr_csv, index=False)
pair_df.to_csv(pair_csv, index=False)

print("\nSaved files:")
print(corr_csv)
print(pair_csv)

print("\n===== Pair summary =====")
display(pair_df)
```

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =========================
# paths
# =========================
ROOT = Path("/Users/yufeizhou/Desktop/heat-exposure-compare")
OUT = ROOT / "outputs/pilot/alphaearth"
OUT.mkdir(parents=True, exist_ok=True)

model_csv = OUT / "city_rawdim_vs_pc_model_summary.csv"
pair_csv = OUT / "city_rawdim_pair_summary.csv"
pc_msg_csv = OUT / "city_pc1_pc4_city_messages.csv"

if not model_csv.exists():
    raise FileNotFoundError(f"Missing file: {model_csv}")
if not pair_csv.exists():
    raise FileNotFoundError(f"Missing file: {pair_csv}")

model_df = pd.read_csv(model_csv).copy()
pair_df = pd.read_csv(pair_csv).copy()

if pc_msg_csv.exists():
    pc_msg_df = pd.read_csv(pc_msg_csv).copy()
else:
    pc_msg_df = pd.DataFrame(columns=["city", "top_city_message"])

print("Loaded:")
print(model_csv)
print(pair_csv)
print(pc_msg_csv if pc_msg_csv.exists() else "city_pc1_pc4_city_messages.csv not found (skip)")

# =========================
# model summary reshape
# =========================
needed_models = [
    "baseline_factors_only",
    "baseline_plus_pc1_to_pc7",
    "baseline_plus_stable_rawdims",
    "baseline_plus_pc1_to_pc7_plus_stable_rawdims",
]

missing_models = sorted(set(needed_models) - set(model_df["model"].unique()))
if missing_models:
    print("WARNING: missing models:", missing_models)

pivot = model_df.pivot_table(
    index="city",
    columns="model",
    values="cv_r2_mean",
    aggfunc="first"
).reset_index()

for m in needed_models:
    if m not in pivot.columns:
        pivot[m] = np.nan

summary = pivot[["city"]].copy()
summary["baseline_r2"] = pivot["baseline_factors_only"]
summary["pc_only_r2"] = pivot["baseline_plus_pc1_to_pc7"]
summary["rawdim_only_r2"] = pivot["baseline_plus_stable_rawdims"]
summary["combined_r2"] = pivot["baseline_plus_pc1_to_pc7_plus_stable_rawdims"]

summary["delta_pc_vs_baseline"] = summary["pc_only_r2"] - summary["baseline_r2"]
summary["delta_rawdim_vs_baseline"] = summary["rawdim_only_r2"] - summary["baseline_r2"]
summary["delta_combined_vs_baseline"] = summary["combined_r2"] - summary["baseline_r2"]
summary["delta_combined_vs_pc"] = summary["combined_r2"] - summary["pc_only_r2"]
summary["delta_combined_vs_rawdim"] = summary["combined_r2"] - summary["rawdim_only_r2"]

def best_model_name(row):
    candidates = {
        "baseline_factors_only": row["baseline_r2"],
        "baseline_plus_pc1_to_pc7": row["pc_only_r2"],
        "baseline_plus_stable_rawdims": row["rawdim_only_r2"],
        "baseline_plus_pc1_to_pc7_plus_stable_rawdims": row["combined_r2"],
    }
    candidates = {k: v for k, v in candidates.items() if pd.notna(v)}
    return max(candidates, key=candidates.get) if len(candidates) else np.nan

summary["best_model"] = summary.apply(best_model_name, axis=1)

def recommendation(row):
    d_pc = row["delta_pc_vs_baseline"]
    d_raw = row["delta_rawdim_vs_baseline"]
    d_combo_pc = row["delta_combined_vs_pc"]

    if pd.isna(d_pc) or pd.isna(d_raw) or pd.isna(d_combo_pc):
        return "Need manual review"

    if d_combo_pc >= 0.02:
        return "Stable raw dims add useful information beyond PCs"
    elif d_raw >= d_pc - 0.03:
        return "Stable raw dims capture similar signal to PCs and help interpretation"
    else:
        return "PC-only model remains strongest; raw dims may mostly overlap with PCs"

summary["recommendation"] = summary.apply(recommendation, axis=1)

# =========================
# top raw-dim / observable pairs
# =========================
pair_df = pair_df.sort_values(["city", "abs_spearman_r"], ascending=[True, False]).copy()

top2 = (
    pair_df.groupby("city", group_keys=False)
    .head(2)
    .copy()
)

def pair_text(sub):
    parts = []
    for _, r in sub.iterrows():
        parts.append(
            f"{r['raw_dim']} ↔ {r['best_observable_var']} (|r|={abs(r['spearman_r']):.3f})"
        )
    return "; ".join(parts)

pair_text_df = (
    top2.groupby("city")
    .apply(pair_text)
    .reset_index(name="top_rawdim_pairs")
)

summary = summary.merge(pair_text_df, on="city", how="left")

# =========================
# merge PC city-level messages if present
# =========================
if len(pc_msg_df) > 0:
    msg_col = None
    for c in ["top_city_message", "pc_summary_short", "city_message", "message"]:
        if c in pc_msg_df.columns:
            msg_col = c
            break

    if msg_col is not None:
        pc_msg_small = pc_msg_df[["city", msg_col]].copy()
        pc_msg_small = pc_msg_small.rename(columns={msg_col: "pc_theme_message"})
        summary = summary.merge(pc_msg_small, on="city", how="left")
    else:
        summary["pc_theme_message"] = np.nan
else:
    summary["pc_theme_message"] = np.nan

# =========================
# round nicer for export
# =========================
num_cols = [
    "baseline_r2",
    "pc_only_r2",
    "rawdim_only_r2",
    "combined_r2",
    "delta_pc_vs_baseline",
    "delta_rawdim_vs_baseline",
    "delta_combined_vs_baseline",
    "delta_combined_vs_pc",
    "delta_combined_vs_rawdim",
]
for c in num_cols:
    summary[c] = summary[c].round(4)

# =========================
# save summary csv
# =========================
summary_csv = OUT / "city_alphaearth_master_summary.csv"
summary.to_csv(summary_csv, index=False)

# =========================
# teacher bullets
# =========================
def pretty_model_name(x):
    mp = {
        "baseline_factors_only": "baseline factors only",
        "baseline_plus_pc1_to_pc7": "baseline + PC1-PC7",
        "baseline_plus_stable_rawdims": "baseline + stable raw dims",
        "baseline_plus_pc1_to_pc7_plus_stable_rawdims": "baseline + PC1-PC7 + stable raw dims",
    }
    return mp.get(x, str(x))

lines = []
lines.append("# AlphaEarth summary for teacher")
lines.append("")
lines.append("This note combines three outputs:")
lines.append("1. PC1-PC4 interpretation")
lines.append("2. Stable raw-dimension vs observable-variable links")
lines.append("3. Predictive comparison of PCs vs stable raw dimensions")
lines.append("")

for _, r in summary.iterrows():
    city = r["city"]
    lines.append(f"## {city}")
    lines.append(
        f"- Model performance (CV R²): baseline={r['baseline_r2']:.4f}, "
        f"PC-only={r['pc_only_r2']:.4f}, rawdim-only={r['rawdim_only_r2']:.4f}, "
        f"combined={r['combined_r2']:.4f}."
    )
    lines.append(
        f"- Gains vs baseline: PC-only={r['delta_pc_vs_baseline']:.4f}, "
        f"rawdim-only={r['delta_rawdim_vs_baseline']:.4f}, "
        f"combined={r['delta_combined_vs_baseline']:.4f}."
    )
    lines.append(
        f"- Best model: {pretty_model_name(r['best_model'])}."
    )
    if pd.notna(r.get("top_rawdim_pairs", np.nan)):
        lines.append(f"- Top raw-dim / observable links: {r['top_rawdim_pairs']}.")
    if pd.notna(r.get("pc_theme_message", np.nan)):
        lines.append(f"- PC interpretation summary: {r['pc_theme_message']}")
    lines.append(f"- Working interpretation: {r['recommendation']}.")
    lines.append("")

lines.append("## Cross-city takeaway")
for _, r in summary.iterrows():
    if r["city"] == "Houston":
        houston_best = pretty_model_name(r["best_model"])
        houston_note = r["recommendation"]
    if r["city"] == "Phoenix":
        phoenix_best = pretty_model_name(r["best_model"])
        phoenix_note = r["recommendation"]

lines.append(
    f"- Houston best model: {houston_best}. "
    f"This suggests {houston_note.lower()}."
)
lines.append(
    f"- Phoenix best model: {phoenix_best}. "
    f"This suggests {phoenix_note.lower()}."
)
lines.append(
    "- Overall, the stable raw dimensions are now interpretable and can be compared directly against the PCA-based summaries."
)

teacher_md = OUT / "city_alphaearth_teacher_bullets.md"
teacher_txt = OUT / "city_alphaearth_teacher_bullets.txt"

teacher_md.write_text("\n".join(lines), encoding="utf-8")
teacher_txt.write_text("\n".join(lines), encoding="utf-8")

# =========================
# cleaner model comparison plot
# =========================
plot_df = summary.copy()
plot_df = plot_df.sort_values("city").reset_index(drop=True)

cities = plot_df["city"].tolist()
x = np.arange(len(cities))
width = 0.18

fig, ax = plt.subplots(figsize=(10, 5))

vals_baseline = plot_df["baseline_r2"].values
vals_pc = plot_df["pc_only_r2"].values
vals_raw = plot_df["rawdim_only_r2"].values
vals_combo = plot_df["combined_r2"].values

ax.bar(x - 1.5 * width, vals_baseline, width, label="Baseline")
ax.bar(x - 0.5 * width, vals_pc, width, label="Baseline + PC1-PC7")
ax.bar(x + 0.5 * width, vals_raw, width, label="Baseline + stable raw dims")
ax.bar(x + 1.5 * width, vals_combo, width, label="Baseline + PC1-PC7 + stable raw dims")

ax.set_xticks(x)
ax.set_xticklabels(cities)
ax.set_ylabel("Cross-validated R²")
ax.set_title("AlphaEarth model comparison by city")
ax.legend()
ax.set_ylim(0, max(np.nanmax(vals_combo), np.nanmax(vals_pc), np.nanmax(vals_raw), np.nanmax(vals_baseline)) + 0.08)

for offset, vals in zip(
    [-1.5 * width, -0.5 * width, 0.5 * width, 1.5 * width],
    [vals_baseline, vals_pc, vals_raw, vals_combo]
):
    for xi, yi in zip(x + offset, vals):
        if pd.notna(yi):
            ax.text(xi, yi + 0.01, f"{yi:.3f}", ha="center", va="bottom", fontsize=8, rotation=90)

plt.tight_layout()

plot_png = OUT / "city_alphaearth_model_comparison_clean.png"
plt.savefig(plot_png, dpi=300, bbox_inches="tight")
plt.show()

# =========================
# display final summary
# =========================
print("\nSaved files:")
print(summary_csv)
print(teacher_md)
print(teacher_txt)
print(plot_png)

print("\n===== Final summary table =====")
display(summary)
```
