# AlphaEarth final summary

This notebook builds a clean meeting-ready summary from notebooks 10–13.

Goals:
1. summarize model performance across cities
2. summarize PC1–PC4 interpretation
3. summarize shared vs city-specific stable raw dimensions
4. summarize whether transferable raw dimensions add value beyond PC1–PC7
5. export clean files for advisor update

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

# =========================================================
# 0) paths
# =========================================================
ROOT = Path("/Users/yufeizhou/Desktop/heat-exposure-compare")
OUT = ROOT / "outputs" / "pilot" / "alphaearth"
OUT.mkdir(parents=True, exist_ok=True)

files = {
    "master_summary": OUT / "city_alphaearth_master_summary.csv",
    "pc_meeting_ready": OUT / "city_pc1_pc4_meeting_ready.csv",
    "shared_summary": OUT / "city_shared_vs_unique_rawdims_summary.csv",
    "transfer_best": OUT / "city_transfer_plus_pc_best_by_city.csv",
    "cross_transfer": OUT / "city_cross_city_transfer_summary.csv",
}

print("==== checking input files ====")
for k, v in files.items():
    print(f"{k}: {v.exists()}  {v}")

missing = [k for k, v in files.items() if not v.exists()]
if missing:
    raise FileNotFoundError(f"Missing required files: {missing}")

# =========================================================
# 1) helpers
# =========================================================
def pick_col(df, candidates, required=True):
    lower_map = {c.lower(): c for c in df.columns}
    for cand in candidates:
        if cand in df.columns:
            return cand
        if cand.lower() in lower_map:
            return lower_map[cand.lower()]
    if required:
        raise KeyError(f"Cannot find any of {candidates} in columns: {list(df.columns)}")
    return None

def pc_sort_value(x):
    s = str(x)
    digits = "".join(ch for ch in s if ch.isdigit())
    return int(digits) if digits else 999

def fmt3(x):
    if pd.isna(x):
        return ""
    return f"{x:.3f}"

# =========================================================
# 2) read inputs
# =========================================================
master_df = pd.read_csv(files["master_summary"]).copy()
pc_df = pd.read_csv(files["pc_meeting_ready"]).copy()
shared_df = pd.read_csv(files["shared_summary"]).copy()
transfer_best_df = pd.read_csv(files["transfer_best"]).copy()
cross_transfer_df = pd.read_csv(files["cross_transfer"]).copy()

print("\n==== input previews ====")
display(master_df.head())
display(pc_df.head())
display(shared_df.head())
display(transfer_best_df.head())
display(cross_transfer_df.head())

# =========================================================
# 3) standardize columns
# =========================================================
master_city_col = pick_col(master_df, ["city"])
baseline_col = pick_col(master_df, ["baseline_r2"])
pc_only_col = pick_col(master_df, ["pc_only_r2"])
rawdim_only_col = pick_col(master_df, ["rawdim_only_r2"])
combined_col = pick_col(master_df, ["combined_r2"])
best_model_col = pick_col(master_df, ["best_model"], required=False)
recommend_col = pick_col(master_df, ["recommendation"], required=False)
top_pairs_col = pick_col(master_df, ["top_rawdim_pairs"], required=False)
pc_theme_msg_col = pick_col(master_df, ["pc_theme_message"], required=False)

pc_city_col = pick_col(pc_df, ["city"])
pc_name_col = pick_col(pc_df, ["pc"])
pc_num_col = pick_col(pc_df, ["pc_num"], required=False)
explained_pct_col = pick_col(pc_df, ["explained_variance_pct"], required=False)
cum_pct_col = pick_col(pc_df, ["cumulative_variance_pct"], required=False)
theme_col = pick_col(pc_df, ["candidate_theme"], required=False)
top_factor_brief_col = pick_col(pc_df, ["top_linked_factors_brief"], required=False)
pc_summary_short_col = pick_col(pc_df, ["pc_summary_short"], required=False)

shared_city_col = pick_col(shared_df, ["city"])
selected_n_col = pick_col(shared_df, ["n_selected_dims"])
shared_n_col = pick_col(shared_df, ["n_shared_dims"])
unique_n_col = pick_col(shared_df, ["n_city_unique_dims"])
shared_dims_col = pick_col(shared_df, ["shared_dims"], required=False)
unique_dims_col = pick_col(shared_df, ["city_unique_dims"], required=False)

transfer_city_col = pick_col(transfer_best_df, ["city"])
transfer_model_col = pick_col(transfer_best_df, ["model"])
transfer_r2_col = pick_col(transfer_best_df, ["cv_r2_mean"], required=False)

cross_city_col = pick_col(cross_transfer_df, ["city"])
cross_model_col = pick_col(cross_transfer_df, ["model"])
cross_r2_col = pick_col(cross_transfer_df, ["cv_r2_mean"], required=False)

# =========================================================
# 4) build PC summary per city
# =========================================================
pc_df = pc_df.copy()
pc_df["_pc_sort"] = pc_df[pc_name_col].map(pc_sort_value)
pc_df = pc_df.sort_values([pc_city_col, "_pc_sort"]).reset_index(drop=True)

pc_summary_map = {}
pc_theme_map = {}
pc_factor_map = {}

for city, sub in pc_df.groupby(pc_city_col):
    lines = []
    theme_lines = []
    factor_lines = []

    for _, row in sub.iterrows():
        pc_name = row[pc_name_col]
        ev = row[explained_pct_col] if explained_pct_col else np.nan
        cum = row[cum_pct_col] if cum_pct_col else np.nan
        theme = row[theme_col] if theme_col else ""
        factors = row[top_factor_brief_col] if top_factor_brief_col else ""
        short = row[pc_summary_short_col] if pc_summary_short_col else ""

        line = f"{pc_name}: {theme}"
        if not pd.isna(ev):
            line += f" ({ev:.1f}%)"
        lines.append(line)

        theme_lines.append(f"{pc_name}={theme}")
        if isinstance(factors, str) and factors.strip():
            factor_lines.append(f"{pc_name}: {factors}")

    pc_summary_map[city] = " | ".join(lines)
    pc_theme_map[city] = " | ".join(theme_lines)
    pc_factor_map[city] = " ; ".join(factor_lines)

# =========================================================
# 5) build final meeting-ready table
# =========================================================
master_df = master_df.copy()
master_df["delta_pc_vs_baseline_calc"] = master_df[pc_only_col] - master_df[baseline_col]
master_df["delta_rawdim_vs_baseline_calc"] = master_df[rawdim_only_col] - master_df[baseline_col]
master_df["delta_combined_vs_baseline_calc"] = master_df[combined_col] - master_df[baseline_col]
master_df["delta_combined_vs_pc_calc"] = master_df[combined_col] - master_df[pc_only_col]
master_df["delta_combined_vs_rawdim_calc"] = master_df[combined_col] - master_df[rawdim_only_col]

rows = []

for _, row in master_df.iterrows():
    city = row[master_city_col]

    shared_row = shared_df.loc[shared_df[shared_city_col] == city]
    if len(shared_row) == 0:
        shared_row = pd.DataFrame([{}])

    transfer_row = transfer_best_df.loc[transfer_best_df[transfer_city_col] == city]
    if len(transfer_row) == 0:
        transfer_row = pd.DataFrame([{}])

    shared_n = shared_row.iloc[0].get(shared_n_col, np.nan)
    unique_n = shared_row.iloc[0].get(unique_n_col, np.nan)
    selected_n = shared_row.iloc[0].get(selected_n_col, np.nan)
    shared_dims = shared_row.iloc[0].get(shared_dims_col, "") if shared_dims_col else ""
    unique_dims = shared_row.iloc[0].get(unique_dims_col, "") if unique_dims_col else ""

    transfer_best_model = transfer_row.iloc[0].get(transfer_model_col, "") if transfer_model_col else ""
    transfer_best_r2 = transfer_row.iloc[0].get(transfer_r2_col, np.nan) if transfer_r2_col else np.nan

    delta_pc = row["delta_pc_vs_baseline_calc"]
    delta_raw = row["delta_rawdim_vs_baseline_calc"]
    delta_combined = row["delta_combined_vs_baseline_calc"]
    delta_combined_vs_pc = row["delta_combined_vs_pc_calc"]

    # simple takeaway
    if delta_combined_vs_pc > 0.02:
        takeaway = (
            f"{city}: stable raw dims add meaningful value beyond PC1–PC7 "
            f"(combined - PC-only = {delta_combined_vs_pc:.3f})."
        )
    elif delta_combined_vs_pc >= -0.01:
        takeaway = (
            f"{city}: PC-only and combined models are very close "
            f"(combined - PC-only = {delta_combined_vs_pc:.3f}); extra raw dims add limited gain."
        )
    else:
        takeaway = (
            f"{city}: adding stable raw dims reduces performance relative to PC-only "
            f"(combined - PC-only = {delta_combined_vs_pc:.3f})."
        )

    rows.append({
        "city": city,
        "baseline_r2": row[baseline_col],
        "pc_only_r2": row[pc_only_col],
        "rawdim_only_r2": row[rawdim_only_col],
        "combined_r2": row[combined_col],
        "delta_pc_vs_baseline": delta_pc,
        "delta_rawdim_vs_baseline": delta_raw,
        "delta_combined_vs_baseline": delta_combined,
        "delta_combined_vs_pc": delta_combined_vs_pc,
        "best_model": row[best_model_col] if best_model_col else "",
        "recommendation": row[recommend_col] if recommend_col else "",
        "top_rawdim_pairs": row[top_pairs_col] if top_pairs_col else "",
        "pc_theme_message": row[pc_theme_msg_col] if pc_theme_msg_col else "",
        "pc1_pc4_summary": pc_summary_map.get(city, ""),
        "pc_linked_factors": pc_factor_map.get(city, ""),
        "n_selected_dims": selected_n,
        "n_shared_dims": shared_n,
        "n_city_unique_dims": unique_n,
        "shared_dims": shared_dims,
        "city_unique_dims": unique_dims,
        "transfer_best_model": transfer_best_model,
        "transfer_best_r2": transfer_best_r2,
        "takeaway": takeaway,
    })

final_df = pd.DataFrame(rows)

final_csv = OUT / "city_alphaearth_final_meeting_table.csv"
final_df.to_csv(final_csv, index=False)

print("\n==== final meeting-ready table ====")
display(final_df)

# =========================================================
# 6) clean final figure
# =========================================================
plot_df = final_df.copy()

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# left: model performance
x = np.arange(len(plot_df))
width = 0.18

axes[0].bar(x - 1.5*width, plot_df["baseline_r2"], width, label="Baseline")
axes[0].bar(x - 0.5*width, plot_df["pc_only_r2"], width, label="Baseline + PC1-PC7")
axes[0].bar(x + 0.5*width, plot_df["rawdim_only_r2"], width, label="Baseline + stable raw dims")
axes[0].bar(x + 1.5*width, plot_df["combined_r2"], width, label="Baseline + PC1-PC7 + stable raw dims")

axes[0].set_xticks(x)
axes[0].set_xticklabels(plot_df["city"])
axes[0].set_ylabel("Cross-validated R²")
axes[0].set_title("AlphaEarth model comparison by city")
axes[0].legend()

for i, col in enumerate(["baseline_r2", "pc_only_r2", "rawdim_only_r2", "combined_r2"]):
    offset = (-1.5 + i) * width
    for j, val in enumerate(plot_df[col]):
        axes[0].text(j + offset, val + 0.01, f"{val:.3f}", ha="center", va="bottom", fontsize=9, rotation=90)

# right: shared vs unique stable dims
axes[1].bar(plot_df["city"], plot_df["n_shared_dims"], label="Shared dims")
axes[1].bar(
    plot_df["city"],
    plot_df["n_city_unique_dims"],
    bottom=plot_df["n_shared_dims"],
    label="City-specific dims"
)
axes[1].set_ylabel("Number of stable raw dimensions")
axes[1].set_title("Shared vs city-specific stable raw dimensions")
axes[1].legend()

for j in range(len(plot_df)):
    total = plot_df.loc[j, "n_shared_dims"] + plot_df.loc[j, "n_city_unique_dims"]
    axes[1].text(j, total + 0.3, f"{int(total)}", ha="center", va="bottom", fontsize=10)

plt.tight_layout()

final_png = OUT / "city_alphaearth_final_overview.png"
plt.savefig(final_png, dpi=300, bbox_inches="tight")
plt.show()

# =========================================================
# 7) teacher-ready note
# =========================================================
bullet_lines = []
bullet_lines.append("# AlphaEarth summary update")
bullet_lines.append("")
bullet_lines.append("This note summarizes the current AlphaEarth results for Houston and Phoenix.")
bullet_lines.append("")

for _, row in final_df.iterrows():
    bullet_lines.append(f"## {row['city']}")
    bullet_lines.append(
        f"- Model performance: baseline={row['baseline_r2']:.3f}, "
        f"baseline+PC1-PC7={row['pc_only_r2']:.3f}, "
        f"baseline+stable raw dims={row['rawdim_only_r2']:.3f}, "
        f"combined={row['combined_r2']:.3f}."
    )
    bullet_lines.append(
        f"- Stable raw dims: selected={int(row['n_selected_dims'])}, "
        f"shared={int(row['n_shared_dims'])}, "
        f"city-specific={int(row['n_city_unique_dims'])}."
    )
    if isinstance(row["shared_dims"], str) and row["shared_dims"].strip():
        bullet_lines.append(f"- Shared stable dims: {row['shared_dims']}.")
    if isinstance(row["top_rawdim_pairs"], str) and row["top_rawdim_pairs"].strip():
        bullet_lines.append(f"- Top raw-dimension pairs: {row['top_rawdim_pairs']}.")
    if isinstance(row["pc1_pc4_summary"], str) and row["pc1_pc4_summary"].strip():
        bullet_lines.append(f"- PC1–PC4 themes: {row['pc1_pc4_summary']}.")
    if isinstance(row["takeaway"], str) and row["takeaway"].strip():
        bullet_lines.append(f"- Takeaway: {row['takeaway']}")
    bullet_lines.append("")

bullet_lines.append("## Cross-city interpretation")
bullet_lines.append("- Shared stable dims are limited in number, while most stable raw dimensions are city-specific.")
bullet_lines.append("- Houston appears to be captured mainly by PC structure; adding stable raw dims gives little extra gain beyond PC1–PC7.")
bullet_lines.append("- Phoenix retains additional value from city-specific stable raw dims beyond PC1–PC7.")
bullet_lines.append("- These results support using both PCA-level interpretation and selected raw-dimension follow-up analysis.")
bullet_lines.append("")

teacher_md = OUT / "city_alphaearth_final_teacher_note.md"
teacher_txt = OUT / "city_alphaearth_final_teacher_note.txt"

teacher_md.write_text("\n".join(bullet_lines), encoding="utf-8")
teacher_txt.write_text("\n".join(bullet_lines), encoding="utf-8")

# =========================================================
# 8) quick terminal-friendly short table
# =========================================================
short_cols = [
    "city",
    "baseline_r2",
    "pc_only_r2",
    "rawdim_only_r2",
    "combined_r2",
    "n_shared_dims",
    "n_city_unique_dims",
    "best_model",
    "takeaway",
]
short_df = final_df[short_cols].copy()

short_csv = OUT / "city_alphaearth_final_short_table.csv"
short_df.to_csv(short_csv, index=False)

print("\n==== saved files ====")
for p in [final_csv, short_csv, final_png, teacher_md, teacher_txt]:
    print(p)

print("\n==== final short table ====")
display(short_df)

print("\n==== teacher-ready note preview ====")
display(Markdown("\n".join(bullet_lines)))
```
